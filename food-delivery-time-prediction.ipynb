#### Outlier Treatment Using IQR Method

```python
cols = ['Delivery_person_Age', 'Delivery_person_Ratings', 'Restaurant_latitude',
        'Restaurant_longitude', 'Delivery_location_longitude', 'multiple_deliveries']

for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower, upper)
```

#### Outlier Treatment Verification

```python
for col in df.select_dtypes(include=['int64', 'float64']).columns.drop('Time_taken(min)'):
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```

#### Separate Features and Target Variable

```python
X = df.drop('Time_taken(min)', axis=1)

y = df['Time_taken(min)']
```

#### One-Hot Encoding and Frequency Encoding

```python
X = pd.get_dummies(
    X,
    columns=['Weatherconditions', 'Road_traffic_density',
             'Type_of_order', 'Type_of_vehicle', 'Festival', 'City'],
    drop_first=True
)

# Frequency Encoding
freq_cols = ['Delivery_person_ID']

freq_mappings = {}

for col in freq_cols:
    freq_map = X[col].value_counts().to_dict()

    # save mapping
    freq_mappings[col] = freq_map

    # apply encoding
    X[col] = X[col].map(freq_map)
```

#### Convert Boolean Features into Numeric Format

```python
bool_cols = X.select_dtypes(include='bool').columns
X[bool_cols] = X[bool_cols].astype(int)
```

#### Split Data into Training and Testing Sets

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split (
    X, y, test_size=0.20, random_state=42
)
```

#### Feature Scaling using StandardScaler

```python
from sklearn.preprocessing import StandardScaler

cols = ['Delivery_person_Age', 'Delivery_person_Ratings', 'Restaurant_latitude', 'Restaurant_longitude',
        'Delivery_location_latitude', 'Delivery_location_longitude', 'Vehicle_condition',
        'multiple_deliveries', 'Order_Day', 'Order_Month', 'Order_Day_of_Week', 'Order_Hour',
        'Order_Minute', 'Pickup_Hour', 'Pickup_Minute']

scaler = StandardScaler()

X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[cols] = scaler.fit_transform(X_train[cols])
X_test_scaled[cols] = scaler.transform(X_test[cols])
```

## MLflow Traking

#### MLflow Tracking Path Setup

```python
import mlflow
import os

mlflow.set_tracking_uri("sqlite:///mlflow.db")
mlflow.set_experiment("Food_Delivery_Prediction")
```

#### Baseline Models Dictionary Setup

```python
models = {
    "Linear Regression": lr_model,
    "KNN Regressor": knn_model,
    "SVM Regressor": svr_model,
    "Decision Tree": dt_model,
    "Random Forest": rf_model,
    "Gradient Boosting": gb_model,
    "XGBoost": xgb_model,
    "LightGBM": lgbm_model,
    "CatBoost": cb_model
}
```

#### MLflow Baseline Model Tracking

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

import mlflow
import mlflow.sklearn
import mlflow.xgboost
import mlflow.lightgbm
import mlflow.catboost

mlflow.set_experiment("Food_Delivery_Prediction")

for model_name, model in models.items():

    with mlflow.start_run(run_name=model_name):

        if model_name in ["Linear Regression", "KNN Regressor", "SVM Regressor"]:
            y_pred = model.predict(X_test_scaled)
        else:
            y_pred = model.predict(X_test)

        mae = mean_absolute_error(y_test, y_pred)
        mse = mean_squared_error(y_test, y_pred)
        rmse = np.sqrt(mse)
        r2 = r2_score(y_test, y_pred)

        mlflow.log_param("Model_Name", model_name)

        mlflow.log_metric("MAE", mae)
        mlflow.log_metric("MSE", mse)
        mlflow.log_metric("RMSE", rmse)
        mlflow.log_metric("R2_Score", r2)

        if model_name == "XGBoost":
            mlflow.xgboost.log_model(model, name="model")

        elif model_name == "LightGBM":
            mlflow.lightgbm.log_model(model, name="model")

        elif model_name == "CatBoost":
            mlflow.catboost.log_model(model, name="model")

        else:
            mlflow.sklearn.log_model(model, name="model")

        print(f"{model_name} tracked successfully")
```
